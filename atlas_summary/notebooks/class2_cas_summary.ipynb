# Class 2 cas summary

Look at the size distribution of class 2 cas from CRISPR-Cas Atlas

```python
import sys, os, re
import pandas as pd
from plotnine import (ggplot, aes, geom_histogram, geom_density, theme_bw, 
                      labs, theme, element_text, scale_x_continuous, geom_vline, 
                      geom_boxplot, geom_jitter, scale_x_discrete, stat_summary,
                      scale_y_continuous, scale_fill_manual, scale_color_manual, geom_hline,
                      element_blank, theme_classic)
```

Filter for evalue

```python
home_dir = os.path.expanduser("~")
data_dir = os.path.join(home_dir, "work", "spacer_analysis", "atlas_summary", "data")
data_file = os.path.join(data_dir, "raw", "class2_cas_raw.csv")
df = pd.read_csv(data_file, header=0)
```

```python
cas13 = df[df["gene_name"].str.startswith("Cas13") & (df["evalue"] == 0)]

p_boxplot = (ggplot(cas13, aes(x="gene_name", y="length", fill="gene_name"))
             + geom_boxplot(outlier_shape="")
            #  + geom_jitter(width=0.1, alpha=0.1)
             + theme_classic()
             + labs(title="Length distribution of Cas13 proteins", x="Cas13 subtype", y="Length (aa)")
             + theme(legend_position="none", axis_text_x=element_text(rotation=45, hjust=1))
            )

display(p_boxplot)
```

```python
cas = df[~df["gene_name"].str.startswith("Cas13") & (df["evalue"] == 0)]

p_boxplot = (ggplot(cas, aes(x="gene_name", y="length", fill="gene_name"))
             + geom_boxplot(outlier_shape="")
            #  + geom_jitter(width=0.1, alpha=0.1)
             + theme_classic()
             + labs(title="Length distribution of Class 2 Cas proteins", x="Class 2 cas subtype", y="Length (aa)")
             + theme(legend_position="none", axis_text_x=element_text(rotation=45, hjust=1))
            )

display(p_boxplot)
```

```python
df_filt = df[df["evalue"] == 0]

p_boxplot = (ggplot(df_filt, aes(x="gene_name", y="length", fill="gene_name"))
             + geom_boxplot(outlier_shape="")
            #  + geom_jitter(width=0.1, alpha=0.1)
             + theme_classic()
             + labs(title="Length distribution of Class 2 Cas proteins", x="Class 2 cas subtype", y="Length (aa)")
             + theme(legend_position="none", axis_text_x=element_text(rotation=45, hjust=1))
            )

display(p_boxplot)
```
